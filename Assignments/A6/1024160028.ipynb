Q1. Consider the following sensor readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
A sensor has an average measurement error of +2°C.
Perform the following without using a for loop:

```python
# a. Add 2°C to every reading using vectorization
import numpy as np
temperature = np.array([25,28,31,35,38,27,33,40])
print(temperature+2)

# b. Convert all temperatures from Celsius to Fahrenheit using:
# F = 9/5*C +32 Formula

f_temp = (9/5*temperature)+32
print(f_temp)  

# c. Identify all readings greater than 32°C using Boolean indexing
ext=temperature[temperature>32]
print(ext)

# d. Count how many readings exceed 32°C
print(len(ext)) # this is my filtered array 

print(np.sum(temperature>32)) 
```

e. Explain how vectorization and Boolean indexing can make data processing more efficient
 than explicitly iterating through every value.

1. C-Level Execution: NumPy arrays are homogeneous (fixed data type) and stored contiguously in memory. Operations run in pre-compiled C code instead of the Python interpreter.
2. SIMD (Single Instruction, Multiple Data): Modern CPUs can execute the same operation on multiple data points simultaneously at the hardware level.
3. Overhead: Python for loops require dynamic type checking and memory lookups on every single iteration, which is orders of magnitude slower

```python
from matplotlib import streamplot
steps = np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
print(steps)
#a
print("Total steps:",np.sum(steps,axis=None))

#b
print("mean of the steps:",round(np.mean(steps)))

#c
print("Max:",np.max(steps))
print("Min:",np.min(steps))

#d
print("sum of the cols:",np.sum(steps,axis=0)) # vertically collapses rows

#e
print("sum of the rows:",np.sum(steps,axis=1)) # horizonally collapses cols

#f
print(np.argmax(steps)) # get index of the max of flattened 1D array (row by row)

## extra info:
# how to get the row,col using argmax ie flatten array.
flat_index = np.argmax(steps)
user_id,day = np.unravel_index(flat_index,steps.shape) 
print("row of the user",user_id)
print("col of the usr:",day)
```

```python
X = np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]])
y = np.array([40,65,30,85])
#a
print("shape X:",X.shape)
print("dim X",X.ndim)

#b
print("Transpose of X\n",X.T) # reshapes the array from 4,3 to 3,4 ie swaps rows and cols

#c
transpose = X.T@X
print("matrix product:\n",transpose) # @ operator performs multiplication.X

#d
inverse = np.linalg.inv(transpose)

#e
OLSE = inverse@X.T@y
print(OLSE)
```
