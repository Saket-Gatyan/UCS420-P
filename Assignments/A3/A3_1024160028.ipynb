```python
#6
data2 = {'Employee ID':[101,102,103,104,105],'Name':['Alice','Bob','Charlie','Diana','Edward'],'Department':['HR','IT','IT','Marketing','Sales'],'Age':[29,34,41,28,38],'Salary':[50000,70000,65000,55000,60000],'Years_of_Experience':[4,8,10,3,12],'Joining_Date':['2020-03-15','2017-07-19','2013-06-01','2021-02-10','2010-11-25'],'Gender':['Female','Male','Male','Female','Male'],'Bonus':[5000,7000,6000,4500,5000],'Rating':[4.5,4.0,3.8,4.7,3.5]}
pf=pd.DataFrame(data2)
pf.head()
#saving the csv file in the current directory
pf.to_csv("employee.csv")

#(a)Shape (number of rows and columns) of the DataFrame.
print("a")
print(pf.shape)
#(b)Summary of the DataFrame that includes the data types and non-null counts for each column
print("b\n")
print(pf.info())
#(c)Generate descriptive statistics
print("c\n")
print(pf.describe())
#d Display the first 5 rows and last 3 rows of the dataset. 
print("d\n")
print(pf.head())
print(pf.tail(n=3))
print("\n")
#e Calculate the following statistics from the dataset
#(i)The average salary of employees
print("average salary of all employees")
print(pf['Salary'].mean())
#(ii) The average salary of employees
print("total bonus of the employee")
print(pf['Bonus'].sum())
#(iii) The youngest employee's age
print("Youngest Employee is",pf['Age'].min(),"years old")
#iv The highest performance rating.
print("Highest Peformer had the rating of",pf['Rating'].max())

#f Sort the DataFrame by the Salary column in descending order. 
#syntax would be same for multiple values
d3=pf.sort_values(by=['Salary'],ascending=False)
print(d3)

#g Add a new column that categorizes employees based on their performance rating

#The values would be added to column based on function.
#creating a function for apply function.
print("\n")
def x(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >=4.0 and rating < 4.5:
        return 'Good for rating'
    elif rating < 4.0:
        return 'Average for rating'

d3['Performance Rating'] = d3['Rating'].apply(x)
print(d3)

#h Identify missing values in the DataFrame
missing_val = d3.isnull().sum()
print(missing_val)

#i Rename the Employee_ID column to ID
d3.rename(columns={'Employee_ID':'ID'},inplace=True)

#j Rename the Employee_ID column to ID
#(i) Have more than 5 years of experience.
exp = d3['Years_of_Experience'] > 5
print("Employees with 5 or more years of experience")
print(exp.sum()) # exp alone print columns with true or false

#(ii) Belong to the IT department.
itp = d3['Department'] =='IT'
print("Total employee in IT")
print(itp.sum())

#k Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary
d3['Tax'] = 0.10*d3['Salary']
print(d3)

#l saving the updated csv file
d3.to_csv("update_employee.csv")

```
