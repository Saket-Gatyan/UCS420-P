```python
# 6.1
def oddSum(n):
  sum = 0
  for i in range(1, n + 1, 2):
    sum += i
  return sum

n = int(input("Enter the value of n: "))
print(oddSum(n))
```

```python
# 6.2
def primeSum(n):
  sum = 0
  for i in range(1, n + 1):
    f = 0
    for j in range(2, i//2 + 1):
      if (i % j == 0):
        f = 1
        break
    if f == 0:
      sum += i
  return sum

n = int(input("Enter the value of n: "))
print(primeSum(n))
```
