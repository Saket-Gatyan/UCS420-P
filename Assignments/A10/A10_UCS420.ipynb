## Q2. NLP Preprocessing Pipeline (Requirement 4)
Implement lowercase conversion, tokenization, stopword removal, and lemmatization to transform raw healthcare queries into normalized tokens for clinical intent analysis.

```python
import re

# Milestone: Text normalization, tokenization, stopword removal & lemmatization
STOP_WORDS = {
    'i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', 'your', 'yours',
    'he', 'him', 'his', 'she', 'her', 'hers', 'it', 'its', 'they', 'them', 'their',
    'am', 'is', 'are', 'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had',
    'do', 'does', 'did', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as',
    'until', 'while', 'of', 'at', 'by', 'for', 'with', 'about', 'against', 'between',
    'into', 'through', 'during', 'before', 'after', 'to', 'from', 'in', 'out', 'on',
    'off', 'then', 'once', 'here', 'there', 'when', 'where', 'why', 'how', 'all',
    'any', 'both', 'each', 'few', 'more', 'some', 'such', 'no', 'nor', 'not', 'only',
    'same', 'so', 'than', 'too', 'very', 'can', 'will', 'just', 'should', 'now',
    'what', 'which', 'who', 'whom', 'this', 'that', 'these', 'those'
}

def lemmatize_token(token):
    if token.endswith('ies') and len(token) > 4:
        return token[:-3] + 'y'
    if token.endswith('es') and len(token) > 4:
        return token[:-2]
    if token.endswith('s') and not token.endswith('ss') and len(token) > 3:
        return token[:-1]
    if token.endswith('ing') and len(token) > 5:
        return token[:-3]
    if token.endswith('ed') and len(token) > 4:
        return token[:-2]
    return token

def preprocess_query(text):
    lowered = text.lower()
    tokens = re.findall(r'\b[a-z]+\b', lowered)
    filtered = [w for w in tokens if w not in STOP_WORDS]
    lemmatized = [lemmatize_token(w) for w in filtered]
    return ' '.join(lemmatized)

# Milestone: Transformation verification on sample query
sample_query = "Patients experiencing irregular palpitations and severe chest pain should seek help."
sample_tokens = re.findall(r'\b[a-z]+\b', sample_query.lower())
sample_filtered = [w for w in sample_tokens if w not in STOP_WORDS]
sample_lemmas = [lemmatize_token(w) for w in sample_filtered]

print("NLP Preprocessing Step-by-Step Breakdown:")
print(f"  1. Raw Query        : {sample_query}")
print(f"  2. Lowercase Tokens : {sample_tokens}")
print(f"  3. Stopwords Removed: {sample_filtered}")
print(f"  4. Lemmatized Tokens: {sample_lemmas}")
print(f"  5. Normalized Output: '{' '.join(sample_lemmas)}'")

# Milestone: Corpus-wide preprocessing execution
corpus_df['cleaned_query'] = corpus_df['query'].apply(preprocess_query)
print("\nCorpus Preprocessing Sample (First 5 Rows):")
print(corpus_df[['intent', 'query', 'cleaned_query']].head().to_string(index=False))
```

## Q3. Feature Extraction, Classifier Training & Performance Evaluation (Requirements 5, 6)
1. Apply TF-IDF for feature extraction across unigrams and bigrams.
2. Train a supervised intent classifier (Multinomial Naive Bayes / Logistic Regression).
3. Evaluate model performance using Accuracy, Precision, Recall, and F1-score.
4. Visualize the confusion matrix across all intent categories.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import numpy as np

# Milestone: TF-IDF feature extraction
tfidf_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X_tfidf = tfidf_vectorizer.fit_transform(corpus_df['cleaned_query'])
y_labels = corpus_df['intent']

# Milestone: Stratified train-test split & classifier training
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y_labels, test_size=0.20, random_state=42, stratify=y_labels
)

intent_classifier = MultinomialNB(alpha=0.1)
intent_classifier.fit(X_train, y_train)
y_pred = intent_classifier.predict(X_test)

# Milestone: Multi-class performance metric calculation
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='macro', zero_division=0)
rec = recall_score(y_test, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_test, y_pred, average='macro', zero_division=0)

metrics_summary = pd.DataFrame({
    "Evaluation Metric": ["Accuracy", "Precision (Macro)", "Recall (Macro)", "F1-Score (Macro)"],
    "Score": [f"{acc * 100:.2f}%", f"{prec * 100:.2f}%", f"{rec * 100:.2f}%", f"{f1 * 100:.2f}%"]
})

print("=" * 60)
print("INTENT CLASSIFIER EVALUATION REPORT")
print("=" * 60)
print(metrics_summary.to_string(index=False))
print("\nDetailed Classification Report:\n")
print(classification_report(y_test, y_pred, zero_division=0))

# Milestone: Confusion matrix visualization
labels_order = sorted(y_labels.unique())
cm = confusion_matrix(y_test, y_pred, labels=labels_order)

plt.figure(figsize=(7, 5))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Intent Classification Confusion Matrix', fontsize=12, pad=12)
plt.colorbar()
tick_marks = np.arange(len(labels_order))
plt.xticks(tick_marks, labels_order, rotation=45, ha='right')
plt.yticks(tick_marks, labels_order)
plt.xlabel('Predicted Intent', fontweight='bold')
plt.ylabel('Ground Truth Intent', fontweight='bold')

thresh = cm.max() / 2.0
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, format(cm[i, j], 'd'),
                 ha="center", va="center",
                 color="white" if cm[i, j] > thresh else "black")

plt.tight_layout()
plt.show()
```

## Q4. Domain Knowledge Base & Cognitive Rules Engine (Requirements 7, 8)
1. Develop a domain-specific knowledge base containing relevant clinical advice and action steps for each intent category.
2. Implement at least 5 cognitive reasoning rules:
   - Rule 1: Emergency Escalation (red-flag triggers like severe chest pain, radiating arm pain)
   - Rule 2: Low-Confidence Handling (confidence threshold < 0.45 triggers clarification request)
   - Rule 3: Unsupported Query Handling (non-cardiology / out-of-domain query detection)
   - Rule 4: Medication Dosage Safety Mandate (prescription disclaimer for drug queries)
   - Rule 5: Critical Vitals Alert (detection of hypertensive crisis or dangerous vital signs)

```python
# Milestone: Domain-specific cardiology knowledge base
CARDIOLOGY_KNOWLEDGE_BASE = {
    "symptoms": {
        "clinical_summary": "Cardiac symptoms indicate potential myocardial ischemia, electrical arrhythmia, or hemodynamic decompensation.",
        "action_plan": "Schedule a clinical evaluation with a cardiologist. Track onset, frequency, duration, and triggers of episodes.",
        "recommendations": "Rest in an upright position. If pain intensifies or is accompanied by dyspnea or syncope, seek immediate emergency care."
    },
    "prevention": {
        "clinical_summary": "Primary prevention focuses on attenuating cardiovascular risk factors like hypertension, hyperlipidemia, and smoking.",
        "action_plan": "Adopt at least 150 minutes of moderate aerobic activity weekly and achieve ideal body weight (BMI 18.5-24.9).",
        "recommendations": "Eliminate tobacco use completely. Maintain blood pressure below 120/80 mmHg and monitor fasting lipid profiles annually."
    },
    "medication": {
        "clinical_summary": "Cardiovascular pharmacotherapy includes antiplatelets, beta blockers, ACE inhibitors, and lipid-lowering statins.",
        "action_plan": "Take prescribed medications consistently at the designated time. Never discontinue or adjust dosages autonomously.",
        "recommendations": "Report persistent dry cough (ACE inhibitors) or uncharacteristic muscle pain (statins) to your physician promptly."
    },
    "diagnosis": {
        "clinical_summary": "Diagnostic tests quantify electrophysiological integrity, cardiac output, ejection fraction, and vascular stenosis.",
        "action_plan": "Standard workup entails a 12-lead ECG, transthoracic echocardiogram, cardiac troponin assay, and exercise stress testing.",
        "recommendations": "Fast for 8-12 hours prior to lipid panels. Avoid caffeine for 24 hours prior to pharmacological or treadmill stress tests."
    },
    "lifestyle": {
        "clinical_summary": "Lifestyle interventions directly modulate arterial stiffness, endothelial function, and metabolic heart health.",
        "action_plan": "Adhere to the DASH diet: restrict sodium to <2,300 mg/day (ideally 1,500 mg/day), emphasize potassium, fiber, and lean proteins.",
        "recommendations": "Prioritize 7-9 hours of restorative sleep nightly to prevent sympathetic nervous system overdrive and blood pressure spikes."
    }
}

# Milestone: Five-tier cognitive reasoning rules engine
EMERGENCY_TRIGGERS = [
    "crushing chest pain", "chest pain radiating", "radiating to left arm", "radiating to arm",
    "radiating to jaw", "heart attack", "unconscious", "collapsed", "gasping for air",
    "severe chest pressure", "cold sweat and nausea"
]

UNSUPPORTED_DOMAIN_KEYWORDS = [
    "tooth", "teeth", "dental", "fracture", "bone", "ankle", "knee", "sprain",
    "hair loss", "skin rash", "dermatology", "python", "code", "flight", "weather"
]

MEDICATION_SAFETY_KEYWORDS = [
    "dosage", "how much", "how many mg", "stop taking", "adjust dose", "double dose", "prescribe"
]

VITALS_REGEX = r'(?:bp|blood pressure)\s*(?:of|is|:)?\s*(\d{2,3})[\s/]+(\d{2,3})'

def evaluate_cognitive_rules(raw_query, predicted_intent, confidence):
    q_lower = raw_query.lower()
    
    # Rule 1: Emergency Escalation
    for trigger in EMERGENCY_TRIGGERS:
        if trigger in q_lower:
            return {
                "rule_triggered": "RULE_1: EMERGENCY_ESCALATION",
                "response": "CRITICAL EMERGENCY ALERT: Your query contains red-flag cardiac symptoms (e.g. crushing chest pressure or radiating pain). Stop all physical activity immediately, sit down, and call 911 or your local emergency response service right now. Do not drive yourself to the hospital.",
                "override": True
            }
            
    # Rule 5: Critical Vitals / Hypertensive Crisis
    bp_match = re.search(VITALS_REGEX, q_lower)
    if bp_match:
        systolic, diastolic = int(bp_match.group(1)), int(bp_match.group(2))
        if systolic >= 180 or diastolic >= 120:
            return {
                "rule_triggered": "RULE_5: HYPERTENSIVE_CRISIS_ALERT",
                "response": f"URGENT VITAL SIGN WARNING: Recorded Blood Pressure ({systolic}/{diastolic} mmHg) meets the clinical criteria for a Hypertensive Crisis. This requires immediate emergency medical evaluation to prevent acute organ damage.",
                "override": True
            }
            
    # Rule 3: Unsupported / Out-of-Domain Query Handling
    for kw in UNSUPPORTED_DOMAIN_KEYWORDS:
        if kw in q_lower:
            return {
                "rule_triggered": "RULE_3: UNSUPPORTED_DOMAIN",
                "response": "OUT-OF-SCOPE ADVISORY: I am a specialized Heart Health Cognitive Assistant. Your query appears to concern a non-cardiovascular medical specialty (e.g. orthopedics, dentistry, or general topics). Please consult an appropriate healthcare specialist for accurate guidance.",
                "override": True
            }
            
    # Rule 2: Low-Confidence Handling
    if confidence < 0.40:
        return {
            "rule_triggered": "RULE_2: LOW_CONFIDENCE_CLARIFICATION",
            "response": "CLARIFICATION NEEDED: Your query is too ambiguous for a reliable clinical assessment. Please rephrase by specifying your exact symptoms, diagnostic tests, medications, or lifestyle questions.",
            "override": True
        }
        
    # Rule 4: Medication Dosage Safety Mandate
    if any(kw in q_lower for kw in MEDICATION_SAFETY_KEYWORDS) and predicted_intent == "medication":
        return {
            "rule_triggered": "RULE_4: MEDICATION_SAFETY_MANDATE",
            "response": "MEDICATION SAFETY MANDATE: Specific drug dosages and therapeutic modifications must be prescribed by your treating physician based on your renal function and clinical history. Never adjust or cease cardiovascular medications autonomously.",
            "override": False
        }
        
    return {"rule_triggered": "STANDARD_INTENT_RETRIEVAL", "response": None, "override": False}

print("Domain Knowledge Base and 5-Tier Cognitive Rules Engine initialized successfully!")
```

## Q5. Integrated Cognitive Assistant & Multi-Scenario Evaluation (Requirements 9, 10 & Tests 1-5)
Integrate NLP preprocessing, TF-IDF feature extraction, intent classification, cognitive rules engine, and knowledge retrieval into a single cognitive assistant class `HeartHealthCognitiveAssistant`. Deploy an interactive multi-query test harness and evaluate the assistant across all required test scenarios:
1. A query belonging to each defined intent (`symptoms`, `prevention`, `medication`, `diagnosis`, `lifestyle`)
2. A differently worded query
3. An unsupported healthcare query
4. A low-confidence query
5. An emergency-related query

```python
# Milestone: End-to-end cognitive assistant integration
class HeartHealthCognitiveAssistant:
    def __init__(self, vectorizer, classifier, knowledge_base):
        self.vectorizer = vectorizer
        self.classifier = classifier
        self.kb = knowledge_base
        
    def process_query(self, query_text):
        cleaned = preprocess_query(query_text)
        vec_input = self.vectorizer.transform([cleaned])
        probs = self.classifier.predict_proba(vec_input)[0]
        max_idx = np.argmax(probs)
        predicted_intent = self.classifier.classes_[max_idx]
        confidence = probs[max_idx]
        
        rule_result = evaluate_cognitive_rules(query_text, predicted_intent, confidence)
        
        if rule_result["override"]:
            final_reply = rule_result["response"]
        else:
            kb_entry = self.kb.get(predicted_intent, {})
            base_reply = f"{kb_entry.get('clinical_summary', '')} Action Plan: {kb_entry.get('action_plan', '')} Lifestyle Note: {kb_entry.get('recommendations', '')}"
            if rule_result["response"]:
                final_reply = f"{rule_result['response']}\n\nGeneral Guidance: {base_reply}"
            else:
                final_reply = base_reply
                
        return {
            "raw_query": query_text,
            "cleaned_query": cleaned,
            "intent": predicted_intent,
            "confidence": confidence,
            "rule": rule_result["rule_triggered"],
            "response": final_reply
        }

# Initialize assistant instance with full corpus training
full_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X_full = full_vectorizer.fit_transform(corpus_df['cleaned_query'])
full_classifier = MultinomialNB(alpha=0.1)
full_classifier.fit(X_full, corpus_df['intent'])

assistant = HeartHealthCognitiveAssistant(full_vectorizer, full_classifier, CARDIOLOGY_KNOWLEDGE_BASE)

# Milestone: Interactive query runner interface
def display_assistant_card(result, test_label=""):
    print("=" * 80)
    if test_label:
        print(f"TEST SCENARIO: {test_label}")
    print("=" * 80)
    print(f"User Query      : \"{result['raw_query']}\"")
    print(f"Cleaned Tokens  : [{result['cleaned_query']}]")
    print(f"Classified Intent: {result['intent'].upper()} (Confidence: {result['confidence'] * 100:.2f}%)")
    print(f"Cognitive Rule  : {result['rule']}")
    print("-" * 80)
    print(f"Assistant Response:\n{result['response']}")
    print("=" * 80 + "\n")

# Milestone: Execution of mandatory test scenarios (Tests 1 to 5)
test_suite = [
    # 1. Queries belonging to each defined intent
    ("1a. Defined Intent - Symptoms", "I am experiencing sharp chest pain and palpitations when climbing stairs"),
    ("1b. Defined Intent - Prevention", "What cardio exercises and habits help prevent coronary artery disease?"),
    ("1c. Defined Intent - Medication", "What are the common side effects and risks of beta blocker pills?"),
    ("1d. Defined Intent - Diagnosis", "How does an ECG or treadmill stress test diagnose blocked arteries?"),
    ("1e. Defined Intent - Lifestyle", "What is the recommended daily sodium limit in the DASH diet for blood pressure?"),
    
    # 2. Differently worded query
    ("2. Differently Worded Query", "How can I keep my ticker running strong and prevent my pipes from clogging?"),
    
    # 3. Unsupported healthcare query
    ("3. Unsupported Healthcare Query", "How should I treat a fractured ankle bone and tooth extraction pain?"),
    
    # 4. Low-confidence query
    ("4. Low-Confidence Query", "I feel somewhat peculiar and strange today"),
    
    # 5. Emergency-related query
    ("5. Emergency-Related Query", "I have crushing chest pain radiating to left arm and cold sweat")
]

print("RUNNING COGNITIVE ASSISTANT EVALUATION SUITE...\n")
for label, query in test_suite:
    res = assistant.process_query(query)
    display_assistant_card(res, test_label=label)
```
