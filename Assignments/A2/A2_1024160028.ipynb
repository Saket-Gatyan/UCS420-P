```python
# 1
roll_no = "1024160015"
L = []
for digit in roll_no:
  L.append(int(digit) * 10)

# i
print(L)

# ii
L.append(25)
print(L)
# .append() adds 25 at the end

L.insert(4, 13)
print(L)
# .insert() adds 13 at 4th index

# iii
L.remove(0)
print(L)
# .remove() removes the first occurrence of 0

L.pop()
print(L)
# .pop() removes the last element in L

# iv
L.sort()
print(L)
L.sort(reverse=True)
print(L)

# v
print(L[:3])
print(L[-1:-4:-1])

# vi
sum = 0
for i in L:
  sum += i
average = sum/len(L)
print(average)
L2 = [x for x in L if x > average]
print(L2)
```

```python
# 2
# i
scores = tuple(L[:8])
print(scores)

# ii
reversed_scores = scores[::-1]
# Cannot reverse tuples in-place unlike lists because tuples are immutable
print(reversed_scores)

# iii
input_score = int(input("Enter a score: "))
print(scores.index(input_score))

# iv
scores[0] = 100
# Tuples are immutable and hence an error it thrown. Lists, on the other hand, are mutable and their values can be modified.
```

```python
first_score, second_score, *remaining_scores = scores
print(first_score)
print(second_score)
print(remaining_scores)
```

```python
# 3
import random as rm
rm.seed(1024160015)

# i
random_numbers = [rm.randint(100, 900) for _ in range(100)]
print(random_numbers)

# ii & iii
odd_count = 0
even_count = 0
for x in random_numbers:
  if (x % 2 == 0):
    even_count += 1
  else:
    odd_count += 1

print(f"Odd count: {odd_count}.")
print(f"Even count: {even_count}.")

# iv
print("\nPRIME NUMBERS")
prime_count = 0
def primeCheck(x):
  f = 0
  for j in range(2, x//2 + 1):
    if (x % j == 0):
      f = 1
      break
  if f == 0:
    return True
  return False

for x in random_numbers:
  if (primeCheck(x)):
    prime_count += 1
    print(x)
print(f"Prime count: {prime_count}.")

prime_random_numbers = [x for x in random_numbers if primeCheck(x)]
print(f"List of prime numbers: {prime_random_numbers}")

# v
max_count = 0
most_frequent = None

for x in random_numbers:
    if random_numbers.count(x) > max_count:
        max_count = random_numbers.count(x)
        most_frequent = x

print(f"Most frequently occurring number: {most_frequent}\nCount: {max_count}")
```

```python
# 4
roll_no = "1024160015"
A = set([])
for digit in roll_no:
  A.add(int(digit) * 7)

B = set([])
for digit in roll_no:
  B.add(int(digit) * 9)

print(f"A: {A}")
print(f"B: {B}")

# vi
print(f"Union of A and B: {A|B}")

# vii
print(f"Intersection of A and B: {A&B}")

# viii
print(f"A - B: {A.difference(B)}")
print(f"B - A: {B.difference(A)}")
# Difference contains elements that belong to the first set but not to the second set. Symmetric difference is basically union - intersection.

# ix
print(f"Symmetric difference: {A.symmetric_difference(B)}")

# x
print(f"A is a subset of B: {A.issubset(B)}")
print(f"B is a superset of A: {B.issuperset(A)}")

# xi
num = int(input("Enter the number to be removed: "))
A.discard(num)
# .discard() handles missing values whereas .remove() throws an error in case of missing values
```

```python
# 5
my_dict = {
"name": "Daksh Aggarwal",
"roll_no": "1024160015",
"branch": "COSE",
"age": 19,
"city": "Hyderabad"
}
print(my_dict)

# i
location = my_dict.pop("city")
my_dict["location"] = location
print(my_dict)

# ii
my_dict['CGPA'] = 'My CGPA'
print(my_dict)

# iii
my_dict['age'] += 1
print(my_dict)

# iv
my_dict_copy = my_dict.copy()
print(my_dict.pop('branch'))
del my_dict_copy['branch']
# .pop() returns the value corresponding to the key being popped. del keyword does not return anything.

# v
print("\nDictionary items:")
for key, value in my_dict.items():
  print(f"{key} -> {value}")

# vi
print("email" in my_dict)
try:
  print(my_dict["email"])
except:
  print("Key does not exist!")

# vii
friend_dict = {
"name": "ABC",
"roll_no": "123",
"branch": "None",
"age": 49,
"city": "Tokyo"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# The dictionary unpacked last (rightmost) wins. In this case, friend_dict's values overwrite my_dict's.

# viii
new_dict = {key: value for key, value in my_dict.items() if isinstance(value, str) and value.replace(' ', '').isalpha()}
print(f"\nNew dictionary: {new_dict}")
```
